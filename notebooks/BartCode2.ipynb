```python
#select only the title and abstract columns and create a copy of the DataFrame which will be used for further processing
dataset = df[['title','abstract']].copy()
```

```python
#rename the columns of the DataFrame to "input" and "output" for further processing
dataset = dataset.rename(columns={
    "abstract":"input",
    "title":"output"
})
```

```python
# Remove missing values 
dataset = dataset.dropna(subset=["input", "output"])

# Remove empty values from the dataset by filtering out rows where either the "input" or "output" column is empty or contains only whitespace
dataset = dataset[
    (dataset["input"].str.strip() != "") &
    (dataset["output"].str.strip() != "")
]

# Check result of the cleaning process by printing the shape of the dataset and the first few rows
print("Dataset shape:", dataset.shape)
#print the first few rows of the cleaned dataset to verify that the cleaning process was successful
print(dataset.head())
```

```python
from sklearn.model_selection import train_test_split

# First: 80% training, 20% temporary
train_data, temp_data = train_test_split(
    dataset,
    test_size=0.20,
    random_state=42
)

# Second: split temporary 20% into 10% validation and 10% test
validation_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42
)

#print the shapes of the training, validation, and test datasets to verify the splits
print("Training:", train_data.shape)
print("Validation:", validation_data.shape)
print("Test:", test_data.shape)
```

```python
#import datasets library to create Hugging Face datasets from pandas DataFrames
from datasets import Dataset

#create hugging face datasets from the pandas DataFrames for traning
train_dataset = Dataset.from_pandas(
    train_data[["input", "output"]].reset_index(drop=True)
)

#create hugging face datasets from the pandas DataFrames for validation
validation_dataset = Dataset.from_pandas(
    validation_data[["input", "output"]].reset_index(drop=True)
)

#create hugging face datasets from the pandas DataFrames for testing
test_dataset = Dataset.from_pandas(
    test_data[["input", "output"]].reset_index(drop=True)
)

#print the training, validation, and test datasets to verify their contents
print(train_dataset)
print(validation_dataset)
print(test_dataset)
```

```python
#load 
from transformers import BartTokenizer

tokenizer = BartTokenizer.from_pretrained(
    "facebook/bart-base"
)

print("Tokenizer loaded successfully")
```

```python
def tokenize_data(batch):

    # Tokenize input (abstract)
    inputs = tokenizer(
        batch["input"],
        max_length=384,
        truncation=True
    )

    # Tokenize output (title)
    targets = tokenizer(
        text_target=batch["output"],
        max_length=64,
        truncation=True
    )

    # Store title tokens as labels
    inputs["labels"] = targets["input_ids"]

    return inputs
```

```python
#create tokenized datasets for training, validation, and testing by applying the tokenize_data function to each dataset
tokenized_train = train_dataset.map(
    tokenize_data,
    batched=True,
    remove_columns=["input", "output"]
)

tokenized_validation = validation_dataset.map(
    tokenize_data,
    batched=True,
    remove_columns=["input", "output"]
)

tokenized_test = test_dataset.map(
    tokenize_data,
    batched=True,
    remove_columns=["input", "output"]
)
```

```python
import torch

predicted_titles = []

batch_size = 2

model.eval()

for i in range(0, len(tokenized_test), batch_size):

    batch = tokenized_test[i:i + batch_size]

    # Automatically pad different length inputs
    inputs = tokenizer.pad(
        {
            "input_ids": batch["input_ids"],
            "attention_mask": batch["attention_mask"]
        },
        padding=True,
        return_tensors="pt"
    )

    input_ids = inputs["input_ids"].to(device)
    attention_mask = inputs["attention_mask"].to(device)

    # Generate summaries/titles
    with torch.no_grad():

        generated_ids = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,
            max_length=64,
            num_beams=2,
            early_stopping=True
        )

    # Convert generated tokens to text
    titles = tokenizer.batch_decode(
        generated_ids,
        skip_special_tokens=True
    )

    predicted_titles.extend(titles)

    # Progress
    if (i + batch_size) % 200 == 0 or i + batch_size >= len(tokenized_test):
        print(
            "Processed:",
            min(i + batch_size, len(tokenized_test)),
            "/",
            len(tokenized_test)
        )

print("\nGeneration completed!")
print("Total predictions:", len(predicted_titles))
```

```python
labels = tokenized_test["labels"]

actual_titles = []

for label in labels:

    label = [
        token if token != -100
        else tokenizer.pad_token_id
        for token in label
    ]

    title = tokenizer.decode(
        label,
        skip_special_tokens=True
    )

    actual_titles.append(title)

print("Total actual titles:", len(actual_titles))
```

```python
from transformers import BartForConditionalGeneration, BartTokenizer
import torch

# Path of your trained model
model_path = r"D:\code\Final Year Project\BART_MODEL\bart-cord19-final"

# Load tokenizer
tokenizer = BartTokenizer.from_pretrained(model_path)

# Load trained BART model
model = BartForConditionalGeneration.from_pretrained(model_path)

# Use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = model.to(device)
model.eval()

print("Model loaded successfully!")
print("Device:", device)
```

```python
def generate_title(input_text):
    
    # Tokenize input
    inputs = tokenizer(
        input_text,
        max_length=512,
        truncation=True,
        padding=True,
        return_tensors="pt"
    )

    # Move input to GPU/CPU
    input_ids = inputs["input_ids"].to(device)
    attention_mask = inputs["attention_mask"].to(device)

    # Generate title
    with torch.no_grad():
        output_ids = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,
            max_length=64,
            num_beams=4,
            early_stopping=True,
            no_repeat_ngram_size=2
        )

    # Convert tokens to text
    title = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )

    return title
```
